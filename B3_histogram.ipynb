```python
!pip install -q opencv-python-headless numpy matplotlib scikit-image pandas

import cv2
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os

from src.utils_cv import baca, tampil, statistik

print("Persiapan B3 berhasil!")
print("OpenCV:", cv2.__version__)
```

```python
daftar_citra = [
    "teh_041.png",  # Normal
    "teh_042.png",  # Gelap
    "teh_043.png",  # Terang
    "teh_044.png",  # Kontras rendah
    "teh_045.png"   # Bayangan
]

plt.figure(figsize=(10, 6))

for nama in daftar_citra:
    img = baca(f"data/sintetis/citra/{nama}")
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    hist = cv2.calcHist([gray], [0], None, [256], [0, 256])

    plt.plot(hist, label=nama)

plt.title("Histogram Citra Daun Teh pada Lima Kondisi")
plt.xlabel("Intensitas Piksel")
plt.ylabel("Jumlah Piksel")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

os.makedirs("hasil", exist_ok=True)
plt.savefig("hasil/b3_histogram_5kondisi.png")
plt.show()
```

```python
# Membaca citra bayangan
img = baca("data/sintetis/citra/teh_045.png")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Fungsi gamma correction
def gamma_correction(citra, gamma):
    tabel = np.array([
        ((i / 255.0) ** (1 / gamma)) * 255
        for i in range(256)
    ]).astype("uint8")
    return cv2.LUT(citra, tabel)

# Berbagai metode peningkatan citra
hasil = {
    "Asli": gray,
    "Negatif": 255 - gray,
    "Cerah +50": cv2.add(gray, 50),
    "Kontras": cv2.convertScaleAbs(gray, alpha=1.5, beta=-40),
    "Regang": cv2.normalize(
        gray, None, 0, 255, cv2.NORM_MINMAX
    ),
    "Gamma 0.5": gamma_correction(gray, 0.5),
    "Gamma 1.8": gamma_correction(gray, 1.8),
    "Ekualisasi": cv2.equalizeHist(gray),
    "CLAHE": cv2.createCLAHE(
        clipLimit=2.0, tileGridSize=(8, 8)
    ).apply(gray)
}

# Menampilkan citra dan histogram
fig, axes = plt.subplots(3, 6, figsize=(18, 10))

for i, (nama, citra) in enumerate(hasil.items()):
    axes.flat[i].imshow(citra, cmap="gray")
    axes.flat[i].set_title(nama)
    axes.flat[i].axis("off")

    hist = cv2.calcHist([citra], [0], None, [256], [0, 256])
    axes.flat[i + 9].plot(hist)
    axes.flat[i + 9].set_title("Histogram " + nama, fontsize=8)

# Menampilkan ukuran data dan simpangan baku
print("Hasil pengujian teh_045.png")

for nama, citra in hasil.items():
    ukuran = citra.nbytes / 1024
    simpangan = np.std(citra)

    print(
        f"{nama:12s} | KB: {ukuran:.2f} | "
        f"Simpangan baku: {simpangan:.2f}"
    )

plt.tight_layout()
plt.savefig("hasil/b3_perbandingan_metode.png")
plt.show()
```

```python
# Membaca citra grayscale
img = baca("data/sintetis/citra/teh_045.png")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Percobaan CLAHE
clip_limits = [1, 2, 4]
tile_sizes = [4, 8, 16]

fig, axes = plt.subplots(3, 3, figsize=(12, 10))

for i, clip in enumerate(clip_limits):
    for j, tile in enumerate(tile_sizes):

        clahe = cv2.createCLAHE(
            clipLimit=clip,
            tileGridSize=(tile, tile)
        )

        hasil_clahe = clahe.apply(gray)

        axes[i, j].imshow(hasil_clahe, cmap="gray")
        axes[i, j].set_title(
            f"Clip={clip}, Tile={tile}\n"
            f"Std={np.std(hasil_clahe):.2f}"
        )
        axes[i, j].axis("off")

plt.tight_layout()
plt.savefig("hasil/b3_eksperimen_clahe.png")
plt.show()
```

```python
# Membaca citra berwarna
img = baca("data/sintetis/citra/teh_045.png")

# Metode 1: CLAHE pada kanal V HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(hsv)

clahe = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(8, 8)
)

v_baru = clahe.apply(v)
hasil_hsv = cv2.merge([h, s, v_baru])
hasil_hsv = cv2.cvtColor(
    hasil_hsv, cv2.COLOR_HSV2BGR
)

# Metode 2: CLAHE pada setiap kanal BGR
b, g, r = cv2.split(img)

b_baru = clahe.apply(b)
g_baru = clahe.apply(g)
r_baru = clahe.apply(r)

hasil_bgr = cv2.merge([
    b_baru, g_baru, r_baru
])

# Menampilkan perbandingan
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Citra Asli")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(hasil_hsv, cv2.COLOR_BGR2RGB))
plt.title("CLAHE pada Kanal V HSV")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(hasil_bgr, cv2.COLOR_BGR2RGB))
plt.title("CLAHE pada Kanal BGR")
plt.axis("off")

plt.tight_layout()
plt.savefig("hasil/b3_clahe_hsv_bgr.png")
plt.show()

print("Simpangan baku kanal V HSV:", np.std(v_baru))
print("Simpangan baku grayscale HSV:",
      np.std(cv2.cvtColor(hasil_hsv, cv2.COLOR_BGR2GRAY)))
print("Simpangan baku grayscale BGR:",
      np.std(cv2.cvtColor(hasil_bgr, cv2.COLOR_BGR2GRAY)))
```

```python
# Fungsi CLAHE pada grayscale
def koreksi_clahe(gray):
    hsv = cv2.cvtColor(
        cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR),
        cv2.COLOR_BGR2HSV
    )
    h, s, v = cv2.split(hsv)

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    v_baru = clahe.apply(v)
    hasil = cv2.merge([h, s, v_baru])

    return cv2.cvtColor(hasil, cv2.COLOR_HSV2BGR)[:, :, 2]


# Daftar citra dan metode koreksi
pengujian = [
    ("Gelap - Berat", "teh_042.png", "cerah"),
    ("Gelap - Ringan", "teh_022.png", "cerah"),
    ("Terang - Berat", "teh_043.png", "gelap"),
    ("Terang - Ringan", "teh_023.png", "gelap"),
    ("Kontras rendah - Berat", "teh_044.png", "clahe"),
    ("Kontras rendah - Ringan", "teh_024.png", "clahe"),
    ("Bayangan - Berat", "teh_045.png", "clahe"),
    ("Bayangan - Ringan", "teh_025.png", "clahe")
]

fig, axes = plt.subplots(8, 2, figsize=(10, 24))

for i, (judul, nama, metode) in enumerate(pengujian):
    img = baca(f"data/sintetis/citra/{nama}")
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    if metode == "cerah":
        hasil_koreksi = cv2.add(gray, 50)

    elif metode == "gelap":
        hasil_koreksi = cv2.convertScaleAbs(
            gray, alpha=1.0, beta=-40
        )

    else:
        hasil_koreksi = koreksi_clahe(gray)

    axes[i, 0].imshow(gray, cmap="gray")
    axes[i, 0].set_title(f"{judul} - Sebelum")
    axes[i, 0].axis("off")

    axes[i, 1].imshow(hasil_koreksi, cmap="gray")
    axes[i, 1].set_title(f"{judul} - Sesudah")
    axes[i, 1].axis("off")

    print(
        nama,
        "| Metode:", metode,
        "| Sebelum:", round(float(np.std(gray)), 2),
        "| Sesudah:", round(float(np.std(hasil_koreksi)), 2)
    )

plt.tight_layout()
plt.savefig("hasil/b3_koreksi_4kondisi.png")
plt.show()
```
